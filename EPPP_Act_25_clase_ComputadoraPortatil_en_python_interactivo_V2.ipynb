```python
class ComputadoraPortatil:
    def __init__(self, marca, bateria, peso):
        self.marca, self.bateria, self.peso = marca, bateria, peso

    def leerdatos(self):
        print(f"\n>> Leyendo registros de {self.marca}...")
        print(f"Batería: {self.bateria} | Peso: {self.peso}")

    def mostrardatos(self):
        print(f"\n--- FICHA TÉCNICA: {self.marca} ---")
        print(f"Marca: {self.marca}\nBatería: {self.bateria}\nPeso: {self.peso}")

def sistema():
    laptops = []
    for i in range(2):
        print(f"\nREGISTRO LAPTOP {i+1}")
        laptops.append(ComputadoraPortatil(input("Marca: "), input("Batería: "), input("Peso: ")))

    while True:
        op = input(f"\nGestionar: 1.{laptops[0].marca} | 2.{laptops[1].marca} | 3.Salir: ")
        if op == "3": break
        if op in ["1", "2"]:
            laptop = laptops[int(op)-1]
            acc = input("Acción: a) Leer datos | b) Mostrar datos: ").lower()
            laptop.leerdatos() if acc == "a" else laptop.mostrardatos() if acc == "b" else print("Error")

if __name__ == "__main__":
    sistema()
    print("\n" + "═"*45 + "\nPrograma terminado con éxito\n" + "═"*45)
    print("programa rializado por Alejandra rivera valenzuela ")
```
